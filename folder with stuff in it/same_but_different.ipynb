```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

db = pd.read_excel("..//resources//2025_01_metropolitan_street.xlsx")
```

Import stuff and database

```python
crimes = db.groupby('LSOA code')['Crime type'].count().reset_index()
crimes.columns = ['LSOA', 'Crimes commited']
crimes
```

```python
crimes = crimes.sort_values(by='Crimes commited',ascending=False)
crimes
```

```python
crimes = crimes.nlargest(10,'Crimes commited')
crimes
```

```python
crime_types = db.groupby('Crime type').size().reset_index(name = "count")
crime_types = crime_types.sort_values(by='count',ascending=False)
crime_types
```
